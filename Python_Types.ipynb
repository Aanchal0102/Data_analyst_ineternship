```python
# Arithmetic Operators (+,-,*,/,%,**)                        # Used with numeric values to perform common mathematical operations

x = 5
y = 8

# print(x+y, x-y, x*y, x/y, x%y, x**y)

print(x+y, x-y, x*y, x/y, x%y, x**y)
```

```python
# Assignment Operators (=,+=,-=,*=,/=,%=,**=)         # Used to assign values to variables

x, y = 5, 8
# print (x=y, x+=y, x-=y, x*=y)

x += y
del print  # Try to delete any existing 'print' variable
print(x)
```

```python


# Comparison Operators (==,!=,<,>,<=,>=)                     # Used to compare two values

# print(x == y, x != y, x < y, x > y, x <= y, x >= y)
x, y = 5, 8

x >= y

result = x >= y
print(result)

result = x <= y
print(result)

result = x == y
print(result)

result = x != y
print(result)
```

```python
# Logical Operators (and, or, not)                                        # Used to combine conditional statements

x, y = 5, 8

print(f"The current value of x is: {x}, and its type is: {type(x)}")
print(f"The current value of y is: {y}, and its type is: {type(y)}")

# print(x and y)

case1 = (x < y) or (x == y)

print(case1, not(case1))
```

```python
# Membership Operators (==,!=,<,>,<=,>=)                     # Used to test if a sequence is presented in an object

# print(x in y, x not in y)

x = [1,2,3,4,5]
y = 6

y not in x
```
