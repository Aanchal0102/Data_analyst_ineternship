## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not. 
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here. 


```python
import pandas as pd

d1 = pd.DataFrame({
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
})

d2 = pd.DataFrame({
    'city': ['Bengalaru', 'Coimbatore', 'Srirangam', 'Pondicherry'],
    'Temperature': [24, 35, 36, 39]
})

d3 = pd.concat([d1, d2])

def is_ut(city):
    """
    Checks if a given city is a Union Territory based on the provided list.
    """
    union_territories = ['Chandigarh', 'Delhi', 'Pondicherry']
    return city in union_territories

# Apply the is_ut function to the 'city' column to create the 'is_ut' column
d3['is_ut'] = d3['city'].apply(is_ut)

print(d3)
```

The temperatures mentioned in ‘Temperature’ column are mentioned in Celsius, we need another column which contains the same in Fahrenheit. 

HINT - 
- Define a function c_to_f which takes input temp in celsius and returns a value with temperature in Fahrenheit.
- To check: c_to_f(10) should return 50. 


```python
# write function here


import pandas as pd

d1 = pd.DataFrame({
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
})

d2 = pd.DataFrame({
    'city': ['Bengalaru', 'Coimbatore', 'Srirangam', 'Pondicherry'],
    'Temperature': [24, 35, 36, 39]
})

d3 = pd.concat([d1, d2])

def is_ut(city):
    union_territories = ['Chandigarh', 'Delhi', 'Pondicherry']
    return city in union_territories

d3['is_ut'] = d3['city'].apply(is_ut)

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    """Converts temperature from Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

# Apply the conversion function to the 'Temperature' column
d3['Temperature_Fahrenheit'] = d3['Temperature'].apply(celsius_to_fahrenheit)

print(d3)


```

## Indexing and selecting rows in DataFrame

Select subset of the dataframe d1 such that it contains the cities which are union territories.

```python
import pandas as pd

d1 = pd.DataFrame({
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
})

# Define the list of Union Territories
union_territories = ['Chandigarh', 'Delhi', 'Pondicherry']

# Create a boolean Series indicating if each city in d1 is a UT
is_ut_d1 = d1['city'].isin(union_territories)

# Select the rows where is_ut_d1 is True
ut_cities_d1 = d1[is_ut_d1]

print("Union Territories in DataFrame d1:")
print(ut_cities_d1)


```

```python
import pandas as pd

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [25, 30, 22, 35, 28],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'occupation': ['Engineer', 'Doctor', 'Artist', 'Chef', 'Teacher']
}

df_personal = pd.DataFrame(data)
print(df_personal) 


```

```python
# Use Groupby of single column with aggregate sum()

import pandas as pd

data = {
    'department': ['Sales', 'Marketing', 'Sales', 'HR', 'Marketing', 'Sales', 'HR'],
    'employee_id': [101, 102, 103, 104, 105, 106, 107],
    'salary': [50000, 60000, 55000, 45000, 65000, 52000, 48000]
}

df = pd.DataFrame(data)

# Group by the 'department' column and calculate the sum of 'salary' for each department
department_salary_sum = df.groupby('department')['salary'].sum()

print("Sum of salaries for each department:")
print(department_salary_sum)
```

```python
# Use Groupby of single column with aggregate count()

import pandas as pd

data = {
    'department': ['Sales', 'Marketing', 'Sales', 'HR', 'Marketing', 'Sales', 'HR'],
    'employee_id': [101, 102, 103, 104, 105, 106, 107],
    'salary': [50000, 60000, 55000, 45000, 65000, 52000, 48000]
}

df = pd.DataFrame(data)

# Group by the 'department' column and count the number of employees in each department
department_employee_count = df.groupby('department')['employee_id'].count()

print("Number of employees in each department:")
print(department_employee_count)
```

```python
# Use Groupby of single column with aggregate min() and max()
import pandas as pd

data = {
    'department': ['Sales', 'Marketing', 'Sales', 'HR', 'Marketing', 'Sales', 'HR'],
    'salary': [50000, 60000, 55000, 45000, 65000, 52000, 48000]
}

df = pd.DataFrame(data)

# Group by 'department' and find the minimum and maximum salary in each department
department_salary_min_max = df.groupby('department')['salary'].agg(['min', 'max'])

print("Minimum and maximum salary for each department:")
print(department_salary_min_max)


```

```python
# Use Groupby of any 2 columns with aggregate mean()

import pandas as pd

data = {
    'department': ['Sales', 'Sales', 'Marketing', 'Marketing', 'HR', 'HR'],
    'region': ['East', 'West', 'North', 'South', 'East', 'West'],
    'salary': [50000, 55000, 60000, 65000, 45000, 48000]
}

df = pd.DataFrame(data)

# Group by 'department' and 'region', then calculate the mean salary for each group
department_region_mean_salary = df.groupby(['department', 'region'])['salary'].mean()

print("Mean salary by department and region:")
print(department_region_mean_salary)

```

```python
# Use Groupby of any 2 columns with aggregate min() and max()


import pandas as pd

data = {
    'department': ['Sales', 'Sales', 'Marketing', 'Marketing', 'HR', 'HR'],
    'region': ['East', 'West', 'North', 'South', 'East', 'West'],
    'salary': [50000, 55000, 60000, 65000, 45000, 48000]
}

df = pd.DataFrame(data)

# Group by 'department' and 'region', then find the minimum and maximum salary in each group
department_region_salary_min_max = df.groupby(['department', 'region'])['salary'].agg(['min', 'max'])

print("Minimum and maximum salary by department and region:")
print(department_region_salary_min_max)
```
