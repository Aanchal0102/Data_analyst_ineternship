# For Loop

*   It is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).
*   Using Break & Contiue keyword



```python
fruits = ["apple", "banana", "cherry",'pineappple','mango']

for i in fruits:
  print(i)
```

```python
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
for x in fruits:
  if x == "cherry":
    continue
  print(x)
```

# While Loop

*   With the while loop we can execute a set of statements as long as a condition is true.
*   Using Break & Contiue keyword

```python
num = 1

while num < 6:
  print(num)
  num += 1
```

```python
num = 1

while num < 6:
  print(num)
  if num == 3:
    break
  num = num + 1
```

```python
num = 0

while num < 10:
  num += 1
  if num == 5:
    continue
  print(num)
```

```python
# Map Function

nums = [1, 2, 3, 4, 5]

def sq(n):
    return n*n

# sq(2)

square = map(sq, nums)

print(list(square))
```

```python
# Zip Function

name = ["Manjeet", "Nikhil", "Shambhavi"]
ID = [4, 1, 3]

mapped = zip(name, ID)

print(list(mapped))
```
