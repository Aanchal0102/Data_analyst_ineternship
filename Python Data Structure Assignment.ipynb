Create colors_set, and fruits_set. (using set() ) and print them

```python
colors = ['red', 'blue', 'orange']

fruits = ['orange', 'grapes', 'apples']

colors_set = set(colors)
fruits_set = set(fruits)


print (colors_set)
print (fruits_set)
```

```python
union_set = colors_set.union(fruits_set)

print (union_set)
```

Find the intersection of both the sets 

```python

colors_set = {'orange', 'blue', 'red'}
fruits_set = {'grapes', 'orange', 'apples'}

intersection_set = colors_set.intersection(fruits_set)

print (intersection_set)
```

count the number of occurences of charachters in string named "sent" using dictionary and print the same.


```python
sent = "Coronavirus Caused Lockdowns Around The World."

# Initialize an empty dictionary to store character counts
char_counts = {}

# Iterate through each character in the string
for char in sent:
    # If the character is already a key in the dictionary, increment its count
    if char in char_counts:
        char_counts[char] += 1
    # Otherwise, add the character as a new key with a count of 1
    else:
        char_counts[char] = 1

# Print the character counts
print("Character counts in the string 'sent':")
for char, count in char_counts.items():
    print(f"'{char}': {count}")
```
